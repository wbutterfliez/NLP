```python
# Install
!pip install tensorflow
!pip install transformers datasets torch

# Imports
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, LSTM, Dense
from transformers import pipeline
```

```python
chatbot = pipeline(
    "text-generation",
    model="microsoft/DialoGPT-medium"
)

# Chat loop
chat_history = ""

while True:
    user_input = input("You: ")

    if user_input.lower() in ["exit", "quit"]:
        break

    chat_history += f"User: {user_input}\nBot:"

    response = chatbot(
        chat_history,
        max_length=200,
        pad_token_id=50256
    )

    reply = response[0]['generated_text'].split("Bot:")[-1].strip()

    print("Bot:", reply)

    chat_history += f" {reply}\n"
```

```python
qa = pipeline("question-answering")

context = """
India is a country in Asia.
New Delhi is the capital of India.
It is known for its diverse culture.
"""

question = "What is the capital of India?"

result = qa(question=question, context=context)

print("Answer:", result['answer'])
print("Score:", result['score'])
```

```python
summarizer = pipeline("summarization")

text = """
Artificial Intelligence (AI) is a branch of computer science that aims
to create intelligent machines. It has applications in healthcare,
education, robotics, and more. AI systems can learn, reason, and improve over time.
"""

summary = summarizer(text, max_length=40, min_length=15, do_sample=False)

print(summary[0]['summary_text'])
```

```python
translator = pipeline("translation_en_to_fr")

text = "Hello, how are you?"

translation = translator(text)

print(translation[0]['translation_text'])
```
